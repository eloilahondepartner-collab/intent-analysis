```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Chargement du fichier Excel mis à jour
file_path = "/Users/eloilahonde/Downloads/Analsye intent/CopyofClosedCasescount.xlsx"
source_df = pd.read_excel(file_path, header=2)
source_df.columns = source_df.columns.str.strip()

# Isoler les tickets réellement routés par l'IA
is_ai_routed = source_df['Routed by AI (Case)'].astype('string').str.strip().str.lower().eq('yes').fillna(False)
df = source_df.loc[is_ai_routed].copy()
df['Is_Reassigned'] = (
    df['[CORE] Teams Updated?? (Case)'].astype('string').str.strip().eq('Yes').fillna(False)
)
df['CSAT_Init_num'] = pd.to_numeric(
    df['Initial CSAT Score (Case)'].astype('string').str.strip().replace({'-': pd.NA, '': pd.NA}),
    errors='coerce'
)
df['Handle_Time_Min'] = (
    pd.to_timedelta(df['Handle Time (Case Clock) (SUM)'], errors='coerce').dt.total_seconds() / 60
)
for col in ['Initial Team (Case)', 'Level 1 - Case Category (Case)', 'Priority (Case)', '[CORE] Country (CF) (Case)']:
    df[col] = df[col].astype('string').str.strip()

print(f"Fichier chargé : {len(source_df):,} lignes au total.")
print(f"Périmètre analysé : {len(df):,} tickets routés par l'IA, {df.shape[1]} colonnes.\n")

# 3. Calcul des KPIs Globaux
total_cases = len(df)
reassigned_cases = df['Is_Reassigned'].sum()
well_routed_cases = total_cases - reassigned_cases
misrouting_rate = (reassigned_cases / total_cases) * 100

ht_well = df[~df['Is_Reassigned']]['Handle_Time_Min'].mean()
ht_reassigned = df[df['Is_Reassigned']]['Handle_Time_Min'].mean()
ht_penalty = ht_reassigned - ht_well
total_hours_lost = (reassigned_cases * ht_penalty) / 60

csat_well = df[~df['Is_Reassigned']]['CSAT_Init_num'].mean()
csat_reassigned = df[df['Is_Reassigned']]['CSAT_Init_num'].mean()
csat_drop = csat_well - csat_reassigned

print("==================================================")
print("       PERFORMANCE DU ROUTAGE AUTOMATIQUE IA      ")
print("==================================================")
print(f"Tickets routés par l'IA        : {total_cases:,.0f}")
print(f"Sans réassignation             : {well_routed_cases:,.0f} ({100-misrouting_rate:.2f}%)")
print(f"Réassignés                     : {reassigned_cases:,.0f} ({misrouting_rate:.2f}%)")
print(f"Handle time moyen, bien routés : {ht_well:.2f} min ({df.loc[~df['Is_Reassigned'], 'Handle_Time_Min'].notna().sum():,} valeurs)")
print(f"Handle time moyen, réassignés  : {ht_reassigned:.2f} min ({df.loc[df['Is_Reassigned'], 'Handle_Time_Min'].notna().sum():,} valeurs)")
print(f"Pénalité AHT estimée           : +{ht_penalty:.2f} min par ticket réassigné")
print(f"Heures-agents perdues estimées : {total_hours_lost:,.1f} h")
print(f"CSAT initial, bien routés      : {csat_well:.2f} / 100 ({df.loc[~df['Is_Reassigned'], 'CSAT_Init_num'].notna().sum():,} valeurs)")
print(f"CSAT initial, réassignés       : {csat_reassigned:.2f} / 100 ({df.loc[df['Is_Reassigned'], 'CSAT_Init_num'].notna().sum():,} valeurs)")
print(f"Écart CSAT (bien - réassignés) : {csat_drop:+.2f} points")
print("==================================================\n")

# 4. Performance par Équipe Initiale (Initial Team)
print("--- TAUS DE MAUVAIS ROUTAGE PAR ÉQUIPE INITIALE ---")
team_kpi = df.groupby('Initial Team (Case)')['Is_Reassigned'].agg(
    Total='count',
    Reassigned='sum',
    Misrouting_Rate=lambda x: (x.sum()/x.count())*100
).reset_index().sort_values(by='Misrouting_Rate', ascending=False)
team_kpi_output = team_kpi.copy()
team_kpi_output['Misrouting_Rate'] = team_kpi_output['Misrouting_Rate'].map(lambda value: f'{value:.2f}%')
team_kpi_output = team_kpi_output.rename(columns={'Initial Team (Case)': 'Équipe IA', 'Total': 'Tickets', 'Reassigned': 'Réassignés', 'Misrouting_Rate': 'Taux de réassignation'})
display(team_kpi_output)

# 5. Focus B2B (Analyse des Vrais vs Faux B2B)
print("\n--- DEEP DIVE B2B ---")
b2b_df = df[df['Initial Team (Case)'] == 'B2B']
b2b_reassigned = b2b_df['Is_Reassigned'].sum()
b2b_rate = (b2b_reassigned / len(b2b_df)) * 100 if len(b2b_df) else float('nan')
b2b_actual = b2b_df['Level 1 - Case Category (Case)'].isin(['B2B', 'Company / Corporate / Enterprise'])
b2b_real = b2b_actual.sum()

print(f"Tickets classés B2B par l'IA          : {len(b2b_df):,}")
print(f"Réassignés parmi ces tickets          : {b2b_reassigned:,} ({b2b_rate:.2f}%)")
print(f"Catégorie finale réellement B2B        : {b2b_real:,} ({b2b_real/len(b2b_df)*100:.2f}% de précision)")
b2b_category = b2b_df['Level 1 - Case Category (Case)']
b2b_missing = b2b_category.isna() | b2b_category.isin(['-', ''])
b2b_labeled = ~b2b_missing
b2b_precision_labeled = b2b_actual[b2b_labeled].mean() * 100 if b2b_labeled.any() else float('nan')
print(f"Catégorie finale manquante / '-'      : {b2b_missing.sum():,}")
print(f"Précision parmi catégories renseignées : {b2b_precision_labeled:.2f}% ({b2b_actual[b2b_labeled].sum():,}/{b2b_labeled.sum():,})")
print("\nCatégories finales des tickets classés B2B (diagnostic des faux positifs) :")
b2b_categories = b2b_df['Level 1 - Case Category (Case)'].value_counts(dropna=False).head(10).rename_axis('Catégorie finale').reset_index(name='Tickets')
display(b2b_categories)

signature_pattern = r'(?i)\b(?:regards|best regards|kind regards|sincerely|thank you|sent from my|cordialement|bien à vous|mit freundlichen grüßen|saludos|un saludo)\b'
b2b_false = b2b_df.loc[~b2b_actual]
signature_marker = b2b_false['Case Details'].astype('string').str.contains(signature_pattern, regex=True, na=False)
print(f"\nFaux B2B contenant un marqueur textuel de signature : {signature_marker.sum():,}/{len(b2b_false):,} ({signature_marker.mean()*100:.2f}%).")
print("Ce repérage par mots-clés est indicatif : il ne prouve pas que la signature a causé le mauvais routage.\n")

# 6. Performance par Priorité
print("\n--- TAUS DE MAUVAIS ROUTAGE PAR PRIORITÉ ---")
priority_kpi = df.groupby('Priority (Case)')['Is_Reassigned'].agg(
    Total='count',
    Reassigned='sum',
    Misrouting_Rate=lambda x: (x.sum()/x.count())*100
).reset_index().sort_values(by='Misrouting_Rate', ascending=False)
priority_kpi_output = priority_kpi.copy()
priority_kpi_output['Misrouting_Rate'] = priority_kpi_output['Misrouting_Rate'].map(lambda value: f'{value:.2f}%')
priority_kpi_output = priority_kpi_output.rename(columns={'Priority (Case)': 'Priorité', 'Total': 'Tickets', 'Reassigned': 'Réassignés', 'Misrouting_Rate': 'Taux de réassignation'})
display(priority_kpi_output)

# 7. Performance par Pays
print("\n--- TAUS DE MAUVAIS ROUTAGE PAR PAYS ---")
country_kpi = df.groupby('[CORE] Country (CF) (Case)')['Is_Reassigned'].agg(
    Total='count',
    Reassigned='sum',
    Misrouting_Rate=lambda x: (x.sum()/x.count())*100
).reset_index().sort_values(by='Total', ascending=False)
focus_countries = ['Germany', 'Belgium', 'France', 'Spain', 'Switzerland']
country_focus = country_kpi.loc[country_kpi['[CORE] Country (CF) (Case)'].isin(focus_countries)].copy()
country_focus['Misrouting_Rate'] = country_focus['Misrouting_Rate'].map(lambda value: f'{value:.2f}%')
country_focus = country_focus.rename(columns={'[CORE] Country (CF) (Case)': 'Pays', 'Total': 'Tickets', 'Reassigned': 'Réassignés', 'Misrouting_Rate': 'Taux de réassignation'})
display(country_focus)

# 8. Diagnostic des causes possibles et recommandations
print("\n--- DIAGNOSTIC ET RECOMMANDATIONS ---")
confidence_columns = [column for column in source_df.columns if 'confidence' in column.lower()]
print(f"Précision B2B observée : {b2b_real:,}/{len(b2b_df):,} ({b2b_real/len(b2b_df)*100:.2f}%).")
print(f"Parmi les faux positifs B2B, {signature_marker.sum():,}/{len(b2b_false):,} ({signature_marker.mean()*100:.2f}%) contiennent un marqueur textuel pouvant évoquer une signature.")
print("Interprétation : c'est un indice à examiner, pas une preuve de causalité; certains marqueurs peuvent aussi apparaître dans le corps du message.")
print("Le seuil actuel de 0.30 et l'asymétrie des prompts sont des éléments du brief, pas des champs mesurés dans cet export.")
print("Aucun score de confiance n'est présent dans les colonnes du fichier; l'effet d'un seuil à 0.55 doit être mesuré sur un jeu de validation étiqueté." if not confidence_columns else f"Colonnes de confiance détectées : {confidence_columns}")
print("Actions recommandées : isoler le corps du message des signatures/citations; appliquer une structure de prompt commune; tester le seuil B2B à 0.55 en surveillant précision et couverture.\n")

b2b_prompt = """Classify the customer's substantive request. Ignore sender names, email addresses, signatures, quoted messages, disclaimers, and ticket metadata. Assign B2B only when the message body explicitly shows a business or corporate need (for example, a company account, business purchase, bulk order, or business invoice/VAT request). A company name, job title, or business-looking email signature alone is not evidence. If the request is a consumer order, delivery, return, refund, product, or store question, select the matching consumer intent instead. Never force B2B when evidence is unclear; choose the best-supported intent or Unclassified. Apply the B2B routing confidence threshold of 0.55 in the routing configuration."""
aftersales_prompt = """Assign Aftersales only to an existing purchase or order: delivery/tracking, missing, wrong or damaged item, return/refund/exchange, warranty/repair, or an order-related invoice/payment issue. Ignore signatures, quoted threads, disclaimers, greetings, and email metadata. Exclude presales, general product/store questions, and business-account requests. Choose the most specific supported intent; if none is clear, use Unclassified."""
print("PROMPT B2B PROPOSÉ (seuil 0.55 à appliquer dans la configuration de routage) :\n" + b2b_prompt + "\n")
print("PROMPT AFTERSALES COMPACT PROPOSÉ (à valider contre le prompt source) :\n" + aftersales_prompt + "\n")

# 9. Génération des graphiques de synthèse
plt.figure(figsize=(10, 5))
sns.barplot(data=team_kpi, x='Initial Team (Case)', y='Misrouting_Rate', hue='Initial Team (Case)', palette='Reds_r', legend=False)
plt.title('Taux de Mauvais Routage (%) par Équipe Initiale')
plt.ylabel('Taux de Réassignation (%)')
plt.xlabel('Équipe Initiale (IA)')
plt.axhline(15, color='green', linestyle='--', label='Cible (< 15%)')
plt.legend()
plt.tight_layout()
plt.savefig('misrouting_by_team.png')
plt.close()

plt.figure(figsize=(10, 5))
sns.barplot(data=country_kpi[country_kpi['[CORE] Country (CF) (Case)'].isin(focus_countries)], x='[CORE] Country (CF) (Case)', y='Misrouting_Rate', hue='[CORE] Country (CF) (Case)', palette='Blues_r', legend=False)
plt.title('Taux de Mauvais Routage (%) par Pays (Markets principaux)')
plt.ylabel('Taux de Réassignation (%)')
plt.xlabel('Pays')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('misrouting_by_country.png')
plt.close()

print("\nAnalyse terminée. Graphiques 'misrouting_by_team.png' et 'misrouting_by_country.png' générés.")
```

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# Analyse autonome du routage IA.
file_path = Path("/Users/eloilahonde/Downloads/Analsye intent/CopyofClosedCasescount.xlsx")
if not file_path.is_file():
    raise FileNotFoundError(f"Fichier Excel introuvable : {file_path}")
source_df = pd.read_excel(file_path, header=2)
source_df.columns = source_df.columns.astype(str).str.strip()
required = ["Routed by AI (Case)", "Initial Team (Case)", "[CORE] Teams Updated?? (Case)", "Handle Time (Case Clock) (SUM)", "Initial CSAT Score (Case)", "Level 1 - Case Category (Case)", "Priority (Case)", "[CORE] Country (CF) (Case)", "Case Details"]
missing = [c for c in required if c not in source_df.columns]
if missing:
    raise KeyError(f"Colonnes requises absentes : {missing}")
is_ai = source_df["Routed by AI (Case)"].astype("string").str.strip().str.casefold().eq("yes").fillna(False)
df_routing = source_df.loc[is_ai].copy()
if df_routing.empty:
    raise ValueError("Aucun ticket marqué Routed by AI = Yes.")
df_routing["Is_Reassigned"] = df_routing["[CORE] Teams Updated?? (Case)"].astype("string").str.strip().str.casefold().eq("yes").fillna(False)
raw_time = df_routing["Handle Time (Case Clock) (SUM)"]
if pd.api.types.is_timedelta64_dtype(raw_time.dtype):
    df_routing["Handle_Time_Min"] = raw_time.dt.total_seconds() / 60
else:
    days = pd.to_numeric(raw_time, errors="coerce")
    durations = pd.to_timedelta(raw_time.where(days.isna()), errors="coerce")
    df_routing["Handle_Time_Min"] = days * 1440
    df_routing.loc[days.isna(), "Handle_Time_Min"] = durations.loc[days.isna()].dt.total_seconds() / 60
df_routing["CSAT_Initial_Num"] = pd.to_numeric(df_routing["Initial CSAT Score (Case)"], errors="coerce")
reassigned = df_routing["Is_Reassigned"]
well = ~reassigned
n = len(df_routing)
n_reassigned = int(reassigned.sum())
rate = reassigned.mean() * 100
ht_well = df_routing.loc[well, "Handle_Time_Min"].mean()
ht_reassigned = df_routing.loc[reassigned, "Handle_Time_Min"].mean()
aht_penalty = ht_reassigned - ht_well
hours_lost = max(0, aht_penalty) * n_reassigned / 60
csat_well = df_routing.loc[well, "CSAT_Initial_Num"].mean()
csat_reassigned = df_routing.loc[reassigned, "CSAT_Initial_Num"].mean()
csat_loss = csat_well - csat_reassigned
fmt = lambda x: "N/D" if pd.isna(x) else f"{x:,.2f}"
pct = lambda x: "N/D" if pd.isna(x) else f"{x:.2f}%"

display(Markdown("## Performance globale du routage IA"))
kpis = pd.DataFrame([
    ("Tickets routés par l'IA", f"{n:,}", "Périmètre des indicateurs."),
    ("Mauvais routage / réassignation", f"{n_reassigned:,} ({pct(rate)})", "Part des tickets dont l'équipe a été mise à jour."),
    ("Handle time moyen — bien routés", f"{fmt(ht_well)} min", f"{df_routing.loc[well, 'Handle_Time_Min'].notna().sum():,} durées exploitables."),
    ("Handle time moyen — réassignés", f"{fmt(ht_reassigned)} min", f"{df_routing.loc[reassigned, 'Handle_Time_Min'].notna().sum():,} durées exploitables."),
    ("Pénalité AHT (réassignés - bien routés)", f"{aht_penalty:+,.2f} min", "Écart moyen par ticket réassigné; négatif = temps inférieur."),
    ("Heures-agent perdues estimées", f"{hours_lost:,.2f} h", "Pénalité AHT positive × volume réassigné."),
    ("CSAT initial moyen — bien routés", f"{fmt(csat_well)} / 100", f"{df_routing.loc[well, 'CSAT_Initial_Num'].notna().sum():,} scores numériques."),
    ("CSAT initial moyen — réassignés", f"{fmt(csat_reassigned)} / 100", f"{df_routing.loc[reassigned, 'CSAT_Initial_Num'].notna().sum():,} scores numériques."),
    ("Écart CSAT (bien routés - réassignés)", f"{csat_loss:+,.2f} points", "Écart descriptif, non ajusté des motifs."),
], columns=["Indicateur", "Valeur", "Lecture business"])
display(kpis)

def routing_rates(data, column, order=None):
    work = data.assign(_group=data[column].astype("string").fillna("Non renseigné").str.strip())
    out = work.groupby("_group", dropna=False).agg(Tickets=("Is_Reassigned", "size"), Reassignes=("Is_Reassigned", "sum"), Taux=("Is_Reassigned", "mean")).reset_index()
    out["Taux mauvais routage (%)"] = out.pop("Taux") * 100
    out = out.rename(columns={"_group": column, "Reassignes": "Réassignés"})
    if order:
        rank = {value.casefold(): i for i, value in enumerate(order)}
        out["_ordre"] = out[column].str.casefold().map(rank).fillna(len(rank))
        out = out.sort_values(["_ordre", "Tickets"], ascending=[True, False]).drop(columns="_ordre")
    else:
        out = out.sort_values("Taux mauvais routage (%)", ascending=False)
    return out.reset_index(drop=True)

display(Markdown("## Taux de mauvais routage par équipe initiale"))
display(routing_rates(df_routing, "Initial Team (Case)"))
display(Markdown("## Diagnostic B2B"))
b2b_mask = df_routing["Initial Team (Case)"].astype("string").str.strip().str.casefold().eq("b2b").fillna(False)
b2b = df_routing.loc[b2b_mask]
if b2b.empty:
    b2b_accuracy = b2b_rate = np.nan
    false_b2b = b2b
    signature_hits = 0
    signature_rate = np.nan
else:
    
    category = b2b["Level 1 - Case Category (Case)"].astype("string").str.strip().str.casefold()
    actual_b2b = category.isin({"b2b", "company / corporate / enterprise"})
    b2b_accuracy = actual_b2b.mean() * 100
    b2b_rate = b2b["Is_Reassigned"].mean() * 100
    false_b2b = b2b.loc[~actual_b2b]
    marker = false_b2b["Case Details"].astype("string").str.contains(re.compile(r"\b(?:regards|best regards|kind regards|sincerely|thank you|sent from my|cordialement|bien à vous|mit freundlichen grüßen|saludos|un saludo)\b", re.I), na=False)
    signature_hits = int(marker.sum())
    signature_rate = marker.mean() * 100 if len(false_b2b) else np.nan
b2b_table = pd.DataFrame([
    ("Tickets initialement classés B2B", f"{len(b2b):,}"),
    ("Taux de réassignation B2B", pct(b2b_rate)),
    ("Précision réelle B2B (catégorie finale B2B ou Company / Corporate / Enterprise)", pct(b2b_accuracy)),
    ("Faux B2B avec marqueur textuel de signature", f"{signature_hits:,}/{len(false_b2b):,} ({pct(signature_rate)})"),
], columns=["Indicateur", "Valeur"])
display(b2b_table)
print("Le marqueur de signature est une heuristique à auditer, pas une preuve de causalité.")
display(Markdown("## Taux de mauvais routage par priorité"))
display(routing_rates(df_routing, "Priority (Case)", ["Low", "Medium", "High", "Immediate"]))
display(Markdown("## Taux de mauvais routage par pays — focus demandé"))
country = routing_rates(df_routing, "[CORE] Country (CF) (Case)")
focus = ["Germany", "Belgium", "France", "Spain", "Switzerland"]
country_focus = country.loc[country["[CORE] Country (CF) (Case)"].str.casefold().isin({x.casefold() for x in focus})].copy()
if country_focus.empty:
    print("Aucun pays ciblé présent parmi les tickets routés par l'IA.")
else:
    country_focus["_ordre"] = country_focus["[CORE] Country (CF) (Case)"].map({x: i for i, x in enumerate(focus)})
    display(country_focus.sort_values("_ordre").drop(columns="_ordre").reset_index(drop=True))
confidence_columns = [c for c in source_df.columns if "confidence" in c.casefold()]
display(Markdown("## Causes racines à confirmer et recommandations"))
causes = pd.DataFrame([
    ("Signatures e-mail", f"{signature_hits:,}/{len(false_b2b):,} faux B2B avec marqueur ({pct(signature_rate)}).", "Retirer signatures/citations et auditer des exemples."),
    ("Seuil de confiance à 0.30", "Valeur rapportée dans le brief; " + (f"champs repérés : {confidence_columns}." if confidence_columns else "aucun score de confiance dans l'export."), "Tester 0.55 sur un jeu étiqueté; suivre précision, couverture et repli."),
    ("Prompts asymétriques", "Hypothèse rapportée dans le brief, non vérifiable dans cet export.", "Harmoniser preuves requises, exclusions et réponse Unclassified."),
], columns=["Piste", "Constat", "Recommandation"])
display(causes)
print("Les signatures sont une piste; le seuil et l'asymétrie doivent être vérifiés dans la configuration en production.")

b2b_prompt = "Classify the customer's substantive request. Ignore sender names, email addresses, signatures, quoted messages, disclaimers, and ticket metadata. Assign B2B only when the message body explicitly shows a business or corporate need (for example, a company account, business purchase, bulk order, or business invoice/VAT request). A company name, job title, or business-looking email signature alone is not evidence. If the request is a consumer order, delivery, return, refund, product, or store question, select the matching consumer intent instead. Never force B2B when evidence is unclear; choose the best-supported intent or Unclassified. Apply the B2B routing confidence threshold of 0.55 in the routing configuration."
after_sales_prompt = "Assign Aftersales only to an existing purchase or order: delivery/tracking, missing, wrong or damaged item, return/refund/exchange, warranty/repair, or an order-related invoice/payment issue. Ignore signatures, quoted threads, disclaimers, greetings, and email metadata. Exclude presales, general product/store questions, and business-account requests. Choose the most specific supported intent; if none is clear, use Unclassified."
print("\nPROMPT B2B OPTIMISÉ (seuil 0.55 à configurer/valider)\n" + b2b_prompt)
print("\nPROMPT AFTERSALES COMPACTÉ\n" + after_sales_prompt)
print("\nAnalyse terminée.")
```

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# 0. Chargement et préparation — les KPI principaux restent au grain ligne d'export
file_path = Path("/Users/eloilahonde/Downloads/Analsye intent/CopyofClosedCasescount.xlsx")
if not file_path.is_file():
    raise FileNotFoundError(f"Fichier introuvable : {file_path}")

source_df = pd.read_excel(file_path, header=2)
source_df.columns = source_df.columns.astype(str).str.strip()

required_columns = [
    "Routed by AI (Case)",
    "Initial Team (Case)",
    "[CORE] Teams Updated?? (Case)",
    "[CORE] Current Teams (Case)",
    "Level 1 - Case Category (Case)",
    "[CORE] Country (CF) (Case)",
    "Priority (Case)",
    "Handle Time (Case Clock) (SUM)",
    "Initial CSAT Score (Case)",
    "Reassignment Macro Usage Count (SUM)",
    "Case Details",
    "Date",
]
missing_columns = [c for c in required_columns if c not in source_df.columns]
if missing_columns:
    raise KeyError(f"Colonnes requises absentes : {missing_columns}")

def clean_label(series):
    values = series.astype("string").str.strip()
    return values.mask(values.isna() | values.eq("") | values.eq("-"), "Non renseigné")

def fmt_num(value, decimals=2):
    return "N/D" if pd.isna(value) else f"{value:,.{decimals}f}"

def fmt_pct(value):
    return "N/D" if pd.isna(value) else f"{value:.2f}%"

ai_mask = (
    source_df["Routed by AI (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)
df_ai = source_df.loc[ai_mask].copy()
if df_ai.empty:
    raise ValueError("Aucune ligne avec 'Routed by AI (Case)' = Yes.")

df_ai["_Export_Row"] = np.arange(len(df_ai))
df_ai["Is_Reassigned"] = (
    df_ai["[CORE] Teams Updated?? (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)
df_ai["_Initial_Team"] = clean_label(df_ai["Initial Team (Case)"])
df_ai["_Current_Team"] = clean_label(df_ai["[CORE] Current Teams (Case)"])
df_ai["_Final_Category"] = clean_label(df_ai["Level 1 - Case Category (Case)"])
df_ai["_Country"] = clean_label(df_ai["[CORE] Country (CF) (Case)"])
df_ai["_Priority"] = clean_label(df_ai["Priority (Case)"])

# Identifiant de dossier extrait du début de Case Details.
# Les calculs KPI restent ligne-à-ligne; l'identifiant sert à l'analyse d'évolution plus bas.
df_ai["_Case_ID"] = (
    df_ai["Case Details"].astype("string").str.extract(r"^\s*(\d+)")[0]
)

# Handle time : durée pandas conservée telle quelle; numérique Excel interprété en jours.
raw_handle_time = df_ai["Handle Time (Case Clock) (SUM)"]
if pd.api.types.is_timedelta64_dtype(raw_handle_time.dtype):
    df_ai["Handle_Time_Min"] = raw_handle_time.dt.total_seconds() / 60
else:
    numeric_days = pd.to_numeric(raw_handle_time, errors="coerce")
    parsed_duration = pd.to_timedelta(
        raw_handle_time.where(numeric_days.isna()), errors="coerce"
    )
    df_ai["Handle_Time_Min"] = numeric_days * 24 * 60
    duration_rows = numeric_days.isna()
    df_ai.loc[duration_rows, "Handle_Time_Min"] = (
        parsed_duration.loc[duration_rows].dt.total_seconds() / 60
    )

df_ai["CSAT_Initial_Num"] = pd.to_numeric(
    df_ai["Initial CSAT Score (Case)"], errors="coerce"
)
df_ai["Reassignment_Macro_Num"] = pd.to_numeric(
    df_ai["Reassignment Macro Usage Count (SUM)"], errors="coerce"
)

total_rows = len(df_ai)
reassigned_mask = df_ai["Is_Reassigned"]
well_routed_mask = ~reassigned_mask
reassigned_rows = int(reassigned_mask.sum())
misrouting_rate = reassigned_mask.mean() * 100

display(Markdown(
    f"## Périmètre\n"
    f"**{total_rows:,} lignes d'export routées par l'IA**; "
    f"**{df_ai['_Case_ID'].nunique():,} dossiers distincts identifiables**. "
    f"Les KPI ci-dessous comptent les lignes, pas les dossiers dédoublonnés."
))

# 1. KPI global de mauvais routage
display(Markdown("## Bloc 1 — Précision et qualité du routage IA"))

global_routing = pd.DataFrame([{
    "Lignes routées par l'IA": f"{total_rows:,}",
    "Lignes réassignées": f"{reassigned_rows:,}",
    "Taux mauvais routage": fmt_pct(misrouting_rate),
    "Lignes bien routées": f"{int(well_routed_mask.sum()):,}",
    "Lignes sans identifiant dossier": f"{int(df_ai['_Case_ID'].isna().sum()):,}",
}])
display(global_routing)

# 2. Taux par équipe initiale — équipes demandées plus toute valeur observée
team_order = [
    "Aftersales", "Marketplace", "Other", "Presales",
    "B2B", "No detection", "Legal", "Services",
]
observed_teams = df_ai["_Initial_Team"].dropna().unique().tolist()
team_order += [team for team in observed_teams if team not in team_order]

team_stats = (
    df_ai.groupby("_Initial_Team", dropna=False)
    .agg(Lignes=("Is_Reassigned", "size"),
         Réassignées=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
    .reindex(team_order)
)
team_stats["Lignes"] = team_stats["Lignes"].fillna(0).astype(int)
team_stats["Réassignées"] = team_stats["Réassignées"].fillna(0).astype(int)
team_stats["Taux mauvais routage"] = team_stats["Taux"].map(fmt_pct)
team_table = team_stats.rename_axis("Équipe initiale").reset_index()
display(Markdown("### Taux de mauvais routage par équipe initiale"))
display(team_table[["Équipe initiale", "Lignes", "Réassignées", "Taux mauvais routage"]])

# 3. Précision et pollution B2B
b2b = df_ai.loc[df_ai["_Initial_Team"].str.casefold().eq("b2b")].copy()
actual_b2b_categories = {"b2b", "company / corporate / enterprise"}
b2b_is_correct = b2b["_Final_Category"].str.casefold().isin(actual_b2b_categories)
b2b_correct_count = int(b2b_is_correct.sum())
b2b_false = b2b.loc[~b2b_is_correct].copy()
b2b_precision = b2b_is_correct.mean() * 100 if len(b2b) else np.nan
b2b_pollution = (len(b2b_false) / len(b2b) * 100) if len(b2b) else np.nan

b2b_summary = pd.DataFrame([
    ("Lignes qualifiées B2B", f"{len(b2b):,}"),
    ("Catégorie finale réellement B2B", f"{b2b_correct_count:,} ({fmt_pct(b2b_precision)})"),
    ("Faux positifs B2B", f"{len(b2b_false):,} ({fmt_pct(b2b_pollution)})"),
], columns=["Indicateur", "Valeur"])
display(Markdown("### Précision B2B et taux de pollution"))
display(b2b_summary)

b2b_pollution_by_category = (
    b2b_false["_Final_Category"].value_counts(dropna=False)
    .rename_axis("Catégorie finale réelle")
    .rename("Lignes")
    .reset_index()
)
if len(b2b_false):
    b2b_pollution_by_category["Part des faux positifs B2B"] = (
        b2b_pollution_by_category["Lignes"] / len(b2b_false) * 100
    ).map(fmt_pct)
    b2b_pollution_by_category["Part de toutes les lignes B2B"] = (
        b2b_pollution_by_category["Lignes"] / len(b2b) * 100
    ).map(fmt_pct)
display(Markdown("### Catégories réelles cachées sous la qualification B2B"))
display(b2b_pollution_by_category)

# 4. Fallback / échappement : Other et No detection
fallback_teams = ["Other", "No detection"]
fallback_rows = df_ai.loc[df_ai["_Initial_Team"].isin(fallback_teams)]
fallback_summary = (
    fallback_rows.groupby("_Initial_Team")
    .agg(Lignes=("Is_Reassigned", "size"),
         Réassignées=("Is_Reassigned", "sum"),
         Taux_réassignation=("Is_Reassigned", "mean"))
    .reindex(fallback_teams)
)
fallback_summary["Lignes"] = fallback_summary["Lignes"].fillna(0).astype(int)
fallback_summary["Réassignées"] = fallback_summary["Réassignées"].fillna(0).astype(int)
fallback_summary["Part du volume IA"] = (
    fallback_summary["Lignes"] / total_rows * 100
).map(fmt_pct)
fallback_summary["Taux de réassignation"] = fallback_summary["Taux_réassignation"].map(fmt_pct)
fallback_summary = fallback_summary.drop(columns="Taux_réassignation")
fallback_summary.loc["Total fallback"] = {
    "Lignes": len(fallback_rows),
    "Réassignées": int(fallback_rows["Is_Reassigned"].sum()),
    "Part du volume IA": fmt_pct(len(fallback_rows) / total_rows * 100),
    "Taux de réassignation": fmt_pct(fallback_rows["Is_Reassigned"].mean() * 100),
}
display(Markdown("### Taux de fallback / échappement"))
display(fallback_summary.rename_axis("Équipe de fallback").reset_index())

# 5–8. AHT, pénalité et gaspillage horaire
handle_well = df_ai.loc[well_routed_mask, "Handle_Time_Min"].mean()
handle_reassigned = df_ai.loc[reassigned_mask, "Handle_Time_Min"].mean()
aht_penalty = handle_reassigned - handle_well
aht_penalty_pct = (
    aht_penalty / handle_well * 100
    if pd.notna(handle_well) and handle_well != 0 else np.nan
)
hours_wasted = reassigned_rows * aht_penalty / 60

aht_summary = pd.DataFrame([
    ("AHT moyen — bien routés", f"{fmt_num(handle_well)} min",
     f"{df_ai.loc[well_routed_mask, 'Handle_Time_Min'].notna().sum():,} lignes avec durée"),
    ("AHT moyen — réassignés", f"{fmt_num(handle_reassigned)} min",
     f"{df_ai.loc[reassigned_mask, 'Handle_Time_Min'].notna().sum():,} lignes avec durée"),
    ("Pénalité AHT par ligne réassignée", f"{aht_penalty:+,.2f} min",
     f"{aht_penalty_pct:+.2f}%" if pd.notna(aht_penalty_pct) else "N/D"),
    ("Heures-agent nettes estimées", f"{hours_wasted:+,.2f} h",
     "Formule demandée : lignes réassignées × pénalité AHT / 60; négatif = gain net."),
], columns=["Indicateur", "Valeur", "Base / interprétation"])
display(Markdown("## Bloc 2 — Efficacité opérationnelle et coût"))
display(aht_summary)

# 9. Résolution directe : zéro macro de réassignation
macro = df_ai["Reassignment_Macro_Num"]
macro_valid = macro.notna() & macro.ge(0)
macro_zero_count = int((macro_valid & macro.eq(0)).sum())
direct_resolution_rate = (
    macro_zero_count / int(macro_valid.sum()) * 100 if macro_valid.any() else np.nan
)
direct_summary = pd.DataFrame([{
    "Lignes à 0 macro": f"{macro_zero_count:,}",
    "Lignes avec macro exploitable": f"{int(macro_valid.sum()):,}",
    "Macros manquantes / invalides": f"{int((~macro_valid).sum()):,}",
    "Taux de résolution directe (sur macros exploitables)": fmt_pct(direct_resolution_rate),
    "Couverture macro": fmt_pct(macro_valid.mean() * 100),
}])
display(Markdown("### Taux de résolution directe"))
display(direct_summary)

# 10. Distribution des transferts (0, 1, 2, 3, 4, 5+)
macro_bucket = pd.Series("Manquant / invalide", index=df_ai.index, dtype="object")
macro_bucket.loc[macro_valid & macro.eq(0)] = "0"
for count in [1, 2, 3, 4]:
    macro_bucket.loc[macro_valid & macro.eq(count)] = str(count)
macro_bucket.loc[macro_valid & macro.ge(5)] = "5+"

bucket_order = ["0", "1", "2", "3", "4", "5+", "Manquant / invalide"]
ping_pong = macro_bucket.value_counts().reindex(bucket_order, fill_value=0).rename_axis(
    "Transferts (macros)"
).rename("Lignes").reset_index()
ping_pong["Part de toutes les lignes export"] = (
    ping_pong["Lignes"] / total_rows * 100
).map(fmt_pct)
ping_pong["Part des macros exploitables"] = ping_pong["Lignes"].map(
    lambda value: fmt_pct(value / int(macro_valid.sum()) * 100)
    if macro_valid.any() and value != 0 else ("0.00%" if macro_valid.any() else "N/D")
)
display(Markdown("### Distribution du ping-pong de réassignation"))
display(ping_pong)

# 11. Impact CSAT initial
csat_well = df_ai.loc[well_routed_mask, "CSAT_Initial_Num"].mean()
csat_reassigned = df_ai.loc[reassigned_mask, "CSAT_Initial_Num"].mean()
csat_drop = csat_well - csat_reassigned
csat_summary = pd.DataFrame([
    ("CSAT initial moyen — bien routés", fmt_num(csat_well),
     f"{df_ai.loc[well_routed_mask, 'CSAT_Initial_Num'].notna().sum():,} scores exploitables"),
    ("CSAT initial moyen — réassignés", fmt_num(csat_reassigned),
     f"{df_ai.loc[reassigned_mask, 'CSAT_Initial_Num'].notna().sum():,} scores exploitables"),
    ("Chute (bien routés - réassignés)", f"{csat_drop:+,.2f} points",
     "Écart descriptif, non ajusté des motifs de contact."),
], columns=["Indicateur", "Valeur", "Base / interprétation"])
display(Markdown("## Bloc 3 — Impact client et analyse business"))
display(csat_summary)

# 12. Taux de mauvais routage par priorité
priority_order = ["Low", "Medium", "High", "Immediate"]
priority_stats = (
    df_ai.groupby("_Priority")
    .agg(Lignes=("Is_Reassigned", "size"),
         Réassignées=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
    .reindex(priority_order + [
        value for value in df_ai["_Priority"].unique()
        if value not in priority_order
    ])
)
priority_stats["Lignes"] = priority_stats["Lignes"].fillna(0).astype(int)
priority_stats["Réassignées"] = priority_stats["Réassignées"].fillna(0).astype(int)
priority_stats["Taux mauvais routage"] = priority_stats["Taux"].map(fmt_pct)
display(Markdown("### Mauvais routage par priorité"))
display(priority_stats.rename_axis("Priorité").reset_index()[
    ["Priorité", "Lignes", "Réassignées", "Taux mauvais routage"]
])

# 13. Taux par tous les pays / marchés observés
country_stats = (
    df_ai.groupby("_Country")
    .agg(Lignes=("Is_Reassigned", "size"),
         Réassignées=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
    .sort_values("Lignes", ascending=False)
)
country_stats["Taux mauvais routage"] = country_stats["Taux"].map(fmt_pct)
display(Markdown("### Mauvais routage par pays / marché — tous les pays observés"))
display(country_stats.rename_axis("Pays / marché").reset_index()[
    ["Pays / marché", "Lignes", "Réassignées", "Taux mauvais routage"]
])

# 14. Focus Suisse et erreurs B2B
swiss = df_ai.loc[df_ai["_Country"].str.casefold().isin(["switzerland", "suisse"])]
swiss_b2b = swiss.loc[swiss["_Initial_Team"].str.casefold().eq("b2b")]
swiss_b2b_correct = swiss_b2b["_Final_Category"].str.casefold().isin(actual_b2b_categories)
swiss_b2b_false = swiss_b2b.loc[~swiss_b2b_correct]
swiss_misrouted = int(swiss["Is_Reassigned"].sum())
swiss_false_b2b_reassigned = int(swiss_b2b_false["Is_Reassigned"].sum())

swiss_b2b_precision = (
    swiss_b2b_correct.mean() * 100 if len(swiss_b2b) else np.nan
)
swiss_b2b_pollution = (
    len(swiss_b2b_false) / len(swiss_b2b) * 100 if len(swiss_b2b) else np.nan
)
swiss_false_b2b_share_of_errors = (
    swiss_false_b2b_reassigned / swiss_misrouted * 100 if swiss_misrouted else np.nan
)
swiss_summary = pd.DataFrame([
    ("Toutes lignes suisses", f"{len(swiss):,}", fmt_pct(swiss["Is_Reassigned"].mean() * 100)),
    ("Lignes suisses réassignées", f"{swiss_misrouted:,}", "—"),
    ("Lignes qualifiées B2B en Suisse", f"{len(swiss_b2b):,}", "—"),
    ("Précision B2B en Suisse", fmt_pct(swiss_b2b_precision), "Catégorie finale B2B ou Company / Corporate / Enterprise"),
    ("Pollution B2B en Suisse", f"{len(swiss_b2b_false):,} ({fmt_pct(swiss_b2b_pollution)})", "Faux B2B / toutes lignes initialement B2B suisses"),
    ("Faux B2B réassignés / toutes erreurs suisses",
     f"{swiss_false_b2b_reassigned:,} ({fmt_pct(swiss_false_b2b_share_of_errors)})",
     "Part des lignes suisses réassignées qui sont aussi de faux B2B"),
], columns=["Indicateur", "Valeur", "Définition"])
display(Markdown("### Focus Suisse & B2B"))
display(swiss_summary)

# 15. Top 15 des flux d'erreurs — un dossier compte une fois par paire d'équipes.
# Le volume principal est le nombre de dossiers distincts; les lignes export sont indiquées aussi.
flow_source = df_ai.loc[
    df_ai["Is_Reassigned"] & df_ai["_Case_ID"].notna()
].copy()
flow_source = flow_source.drop_duplicates(
    subset=["_Case_ID", "_Initial_Team", "_Current_Team"]
)
top_flows = (
    flow_source.groupby(["_Initial_Team", "_Current_Team"])
    .agg(Volume=("_Case_ID", "nunique"))
    .reset_index()
    .rename(columns={
        "_Initial_Team": "Équipe initiale",
        "_Current_Team": "Équipe courante observée",
    })
)
line_counts = (
    df_ai.loc[df_ai["Is_Reassigned"]]
    .groupby(["_Initial_Team", "_Current_Team"])
    .size()
    .rename("Lignes export")
    .reset_index()
    .rename(columns={
        "_Initial_Team": "Équipe initiale",
        "_Current_Team": "Équipe courante observée",
    })
)
top_flows = top_flows.merge(
    line_counts, on=["Équipe initiale", "Équipe courante observée"], how="left"
)
display(Markdown(
    "## Bloc 4 — Matrice de confusion et flux croisés\n"
    "### Top 15 des flux d'erreurs (volume = dossiers distincts par paire)"
))
print(top_flows.sort_values(by="Volume", ascending=False).head(15).to_string(index=False))

# 16. Alignement complet Équipe initiale × Catégorie réelle finale
taxonomy_counts = pd.crosstab(
    df_ai["_Initial_Team"],
    df_ai["_Final_Category"],
    dropna=False,
).reindex(team_order, fill_value=0)
taxonomy_counts.index.name = "Équipe initiale IA"
taxonomy_counts.columns.name = "Catégorie réelle finale"
display(Markdown("### Matrice complète des volumes — équipe initiale × catégorie finale"))
display(taxonomy_counts)

taxonomy_pct = taxonomy_counts.div(
    taxonomy_counts.sum(axis=1).replace(0, np.nan), axis=0
) * 100
taxonomy_pct = taxonomy_pct.applymap(
    lambda value: fmt_pct(value) if pd.notna(value) else "N/D"
)
display(Markdown("### Même matrice en % au sein de chaque équipe initiale"))
display(taxonomy_pct)

# Analyse complémentaire demandée : évolution des valeurs observées par dossier.
# Date est une période hebdomadaire dans l'export; l'ordre est donc par période,
# sans prétendre établir l'ordre exact des transferts à l'intérieur d'une semaine.
start_text = (
    df_ai["Date"].astype("string")
    .str.extract(r"([A-Za-z]+\s+\d{1,2}\s*,?\s+\d{4})")[0]
    .str.replace(r"\s+,", ",", regex=True)
)
df_ai["_Period_Start"] = pd.to_datetime(
    start_text, format="%B %d, %Y", errors="coerce"
)
df_ai["_Period_Label"] = df_ai["_Period_Start"].dt.strftime("%Y-%m-%d")
df_ai["_Period_Label"] = df_ai["_Period_Label"].fillna("Période non analysable")

def unique_labels(series):
    return list(pd.unique(series.dropna().astype(str)))

def summarize_case(case_id, group):
    ordered = group.sort_values(
        ["_Period_Start", "_Export_Row"], na_position="last"
    )
    timeline = []
    for period, period_rows in ordered.groupby("_Period_Label", sort=False):
        teams = unique_labels(period_rows["_Current_Team"])
        timeline.append(f"{period}: {' / '.join(teams)}")
    current_teams = unique_labels(group["_Current_Team"])
    initial_teams = unique_labels(group["_Initial_Team"])
    final_categories = unique_labels(group["_Final_Category"])
    macros = group["Reassignment_Macro_Num"].dropna()

    return {
        "ID dossier": case_id,
        "Lignes export": len(group),
        "Périodes observées": group["_Period_Label"].nunique(),
        "Équipes courantes distinctes": len(current_teams),
        "Équipes initiales observées": " / ".join(initial_teams),
        "Équipes courantes observées": " / ".join(current_teams),
        "Catégories finales observées": " / ".join(final_categories),
        "Réassigné sur au moins une ligne": bool(group["Is_Reassigned"].any()),
        "Maximum macros de réassignation observé": macros.max() if len(macros) else np.nan,
        "Évolution par période": " ; ".join(timeline),
    }

case_rows = [
    summarize_case(case_id, group)
    for case_id, group in df_ai.loc[df_ai["_Case_ID"].notna()].groupby("_Case_ID", sort=False)
]
case_evolution = pd.DataFrame(case_rows)

multi_team_cases = int((case_evolution["Équipes courantes distinctes"] > 1).sum())
case_evolution_summary = pd.DataFrame([
    ("Dossiers avec ID exploitable", f"{len(case_evolution):,}"),
    ("Lignes IA sans ID exploitable", f"{int(df_ai['_Case_ID'].isna().sum()):,}"),
    ("Dossiers présents sur plusieurs lignes d'export",
     f"{int((case_evolution['Lignes export'] > 1).sum()):,}"),
    ("Dossiers avec plusieurs équipes courantes observées",
     f"{multi_team_cases:,} ({fmt_pct(multi_team_cases / len(case_evolution) * 100) if len(case_evolution) else 'N/D'})"),
], columns=["Indicateur dossier", "Valeur"])
display(Markdown("### Évolution complémentaire groupée par dossier"))
display(case_evolution_summary)

team_evolution_distribution = (
    case_evolution["Équipes courantes distinctes"].value_counts()
    .sort_index()
    .rename_axis("Nombre d'équipes courantes observées")
    .rename("Dossiers")
    .reset_index()
)
team_evolution_distribution["Part des dossiers avec ID"] = (
    team_evolution_distribution["Dossiers"] / len(case_evolution) * 100
).map(fmt_pct)
display(Markdown("### Nombre d'équipes courantes observées par dossier"))
display(team_evolution_distribution)

most_evolving_cases = case_evolution.sort_values(
    ["Équipes courantes distinctes", "Périodes observées", "Lignes export"],
    ascending=False,
).head(20)
display(Markdown(
    "### 20 dossiers les plus évolutifs — équipes courantes observées par période"
))
display(most_evolving_cases)

print(
    "\nNote d'interprétation : les KPI opérationnels ci-dessus comptent les lignes "
    "de l'export. La vue dossier sert à révéler répétitions et changements d'équipes; "
    "les périodes sont hebdomadaires, donc elles ne prouvent pas l'ordre exact des transferts."
)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# Chargement demandé : header=0
file_path = Path("/Users/eloilahonde/Downloads/Analsye intent/CopyofClosedCasescount.xlsx")
if not file_path.is_file():
    raise FileNotFoundError(f"Fichier introuvable : {file_path}")

source_df = pd.read_excel(file_path, header=2)
source_df.columns = source_df.columns.astype(str).str.strip()

required = [
    "Routed by AI (Case)", "Initial Team (Case)",
    "[CORE] Teams Updated?? (Case)", "[CORE] Current Teams (Case)",
    "Level 1 - Case Category (Case)", "[CORE] Country (CF) (Case)",
    "Priority (Case)", "Handle Time (Case Clock) (SUM)",
    "Initial CSAT Score (Case)", "Reassignment Macro Usage Count (SUM)",
    "Case Details", "Date",
]
missing = [c for c in required if c not in source_df.columns]
if missing:
    raise KeyError(
        f"Colonnes absentes avec header=0 : {missing}. "
        f"En-têtes lus : {source_df.columns.tolist()}"
    )

def clean_label(series):
    s = series.astype("string").str.strip()
    return s.mask(s.isna() | s.eq("") | s.eq("-"), "Non renseigné")

def fmt_num(x, digits=2):
    return "N/D" if pd.isna(x) else f"{x:,.{digits}f}"

def fmt_pct(x):
    return "N/D" if pd.isna(x) else f"{x:.2f}%"

# Filtre des lignes routées par l'IA; les KPI ligne restent au grain export.
ai_mask = (
    source_df["Routed by AI (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)
df_line = source_df.loc[ai_mask].copy()
if df_line.empty:
    raise ValueError("Aucune ligne avec Routed by AI (Case) = Yes.")

df_line["_Export_Row"] = np.arange(len(df_line))
df_line["Is_Reassigned"] = (
    df_line["[CORE] Teams Updated?? (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)
df_line["_Initial_Team"] = clean_label(df_line["Initial Team (Case)"])
df_line["_Current_Team"] = clean_label(df_line["[CORE] Current Teams (Case)"])
df_line["_Final_Category"] = clean_label(df_line["Level 1 - Case Category (Case)"])
df_line["_Country"] = clean_label(df_line["[CORE] Country (CF) (Case)"])
df_line["_Priority"] = clean_label(df_line["Priority (Case)"])
df_line["_Case_ID"] = (
    df_line["Case Details"].astype("string").str.extract(r"^\s*(\d+)")[0]
)

# Handle time : les durées pandas sont converties directement; les nombres Excel sont des jours.
raw_time = df_line["Handle Time (Case Clock) (SUM)"]
if pd.api.types.is_timedelta64_dtype(raw_time.dtype):
    df_line["Handle_Time_Min"] = raw_time.dt.total_seconds() / 60
else:
    numeric_days = pd.to_numeric(raw_time, errors="coerce")
    parsed_duration = pd.to_timedelta(
        raw_time.where(numeric_days.isna()), errors="coerce"
    )
    df_line["Handle_Time_Min"] = numeric_days * 24 * 60
    is_duration = numeric_days.isna()
    df_line.loc[is_duration, "Handle_Time_Min"] = (
        parsed_duration.loc[is_duration].dt.total_seconds() / 60
    )

df_line["CSAT_Initial"] = pd.to_numeric(
    df_line["Initial CSAT Score (Case)"], errors="coerce"
)
df_line["Reassign_Macros"] = pd.to_numeric(
    df_line["Reassignment Macro Usage Count (SUM)"], errors="coerce"
)

# Date est un intervalle hebdomadaire; on extrait sa date de début.
date_start = (
    df_line["Date"].astype("string")
    .str.extract(r"([A-Za-z]+\s+\d{1,2}\s*,?\s+\d{4})")[0]
    .str.replace(r"\s+,", ",", regex=True)
)
df_line["_Period_Start"] = pd.to_datetime(date_start, errors="coerce")
df_line["_Period_Label"] = df_line["_Period_Start"].dt.strftime("%Y-%m-%d")
df_line["_Period_Label"] = df_line["_Period_Label"].fillna("Période non analysable")

# Grain dossier : première équipe et dernière équipe/catégorie selon période puis ordre export.
def first_real(series):
    vals = [str(x) for x in series if pd.notna(x) and str(x) != "Non renseigné"]
    return vals[0] if vals else "Non renseigné"

def last_real(series):
    vals = [str(x) for x in series if pd.notna(x) and str(x) != "Non renseigné"]
    return vals[-1] if vals else "Non renseigné"

case_rows = []
ordered = df_line.sort_values(
    ["_Case_ID", "_Period_Start", "_Export_Row"],
    na_position="last",
    kind="stable",
)
for case_id, group in ordered.loc[ordered["_Case_ID"].notna()].groupby(
    "_Case_ID", sort=False
):
    group = group.sort_values(
        ["_Period_Start", "_Export_Row"], na_position="last", kind="stable"
    )
    aht_values = group["Handle_Time_Min"].dropna()
    macro_values = group["Reassign_Macros"].dropna()
    csat_values = group["CSAT_Initial"].dropna()

    case_rows.append({
        "_Case_ID": case_id,
        "_Initial_Team": first_real(group["_Initial_Team"]),
        "_Current_Team": last_real(group["_Current_Team"]),
        "_Final_Category": last_real(group["_Final_Category"]),
        "_Country": first_real(group["_Country"]),
        "_Priority": first_real(group["_Priority"]),
        "Is_Reassigned": bool(group["Is_Reassigned"].any()),
        "Handle_Time_Min": aht_values.max() if len(aht_values) else np.nan,
        "Reassign_Macros": macro_values.max() if len(macro_values) else np.nan,
        "CSAT_Initial": csat_values.iloc[0] if len(csat_values) else np.nan,
        "Export_Rows": len(group),
    })

df_case = pd.DataFrame(case_rows)
if df_case.empty:
    raise ValueError("Aucun ID dossier numérique n'a été extrait de Case Details.")

n_lines = len(df_line)
n_line_reassigned = int(df_line["Is_Reassigned"].sum())
line_misrouting = df_line["Is_Reassigned"].mean() * 100
n_cases = len(df_case)
n_case_reassigned = int(df_case["Is_Reassigned"].sum())
case_misrouting = df_case["Is_Reassigned"].mean() * 100

display(Markdown("## Périmètre et taux global de mauvais routage"))
display(pd.DataFrame([
    ("Lignes IA", f"{n_lines:,}", fmt_pct(line_misrouting)),
    ("Dossiers IA avec ID extrait", f"{n_cases:,}", fmt_pct(case_misrouting)),
    ("Lignes IA sans ID dossier exploitable",
     f"{int(df_line['_Case_ID'].isna().sum()):,}", "Exclues des KPI au grain dossier"),
], columns=["Grain", "Volume", "Taux mauvais routage / note"]))

# BLOC 1 — équipes, B2B, fallback
display(Markdown("## Bloc 1 — Précision et qualité du routage IA"))

team_order = [
    "Aftersales", "Marketplace", "Other", "Presales",
    "B2B", "No detection", "Legal", "Services",
]
team_order += [
    x for x in df_case["_Initial_Team"].unique() if x not in team_order
]

def routing_by_team(data, grain_label):
    stats = (
        data.groupby("_Initial_Team")
        .agg(Lignes=("Is_Reassigned", "size"),
             Réassignées=("Is_Reassigned", "sum"),
             Taux=("Is_Reassigned", "mean"))
        .reindex(team_order)
    )
    stats["Lignes"] = stats["Lignes"].fillna(0).astype(int)
    stats["Réassignées"] = stats["Réassignées"].fillna(0).astype(int)
    stats["Taux mauvais routage"] = stats["Taux"].map(fmt_pct)
    stats["Grain"] = grain_label
    return stats.rename_axis("Équipe initiale").reset_index()[
        ["Grain", "Équipe initiale", "Lignes", "Réassignées", "Taux mauvais routage"]
    ]

display(Markdown("### Taux par équipe initiale — dossier et ligne d'export"))
display(pd.concat([
    routing_by_team(df_case, "Dossier"),
    routing_by_team(df_line, "Ligne export"),
], ignore_index=True))

valid_b2b_categories = {"b2b", "company / corporate / enterprise"}
b2b_cases = df_case.loc[df_case["_Initial_Team"].str.casefold().eq("b2b")]
b2b_correct = b2b_cases["_Final_Category"].str.casefold().isin(valid_b2b_categories)
b2b_false = b2b_cases.loc[~b2b_correct]
b2b_precision = b2b_correct.mean() * 100 if len(b2b_cases) else np.nan
b2b_pollution = len(b2b_false) / len(b2b_cases) * 100 if len(b2b_cases) else np.nan

display(Markdown("### Précision B2B et pollution par catégories réelles"))
display(pd.DataFrame([
    ("Dossiers qualifiés B2B", f"{len(b2b_cases):,}"),
    ("Catégorie finale réellement B2B / Corporate",
     f"{int(b2b_correct.sum()):,} ({fmt_pct(b2b_precision)})"),
    ("Faux positifs B2B", f"{len(b2b_false):,} ({fmt_pct(b2b_pollution)})"),
], columns=["Indicateur", "Valeur"]))

b2b_pollution_categories = (
    b2b_false["_Final_Category"].value_counts()
    .rename_axis("Catégorie finale réelle")
    .rename("Dossiers")
    .reset_index()
)
if len(b2b_false):
    b2b_pollution_categories["Part des faux B2B"] = (
        b2b_pollution_categories["Dossiers"] / len(b2b_false) * 100
    ).map(fmt_pct)
    b2b_pollution_categories["Part de tous les dossiers B2B"] = (
        b2b_pollution_categories["Dossiers"] / len(b2b_cases) * 100
    ).map(fmt_pct)
display(b2b_pollution_categories)

fallback_teams = ["Other", "No detection"]
fallback = df_case.loc[df_case["_Initial_Team"].isin(fallback_teams)]
fallback_stats = (
    fallback.groupby("_Initial_Team")
    .agg(Dossiers=("Is_Reassigned", "size"),
         Réassignés=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
    .reindex(fallback_teams)
)
fallback_stats["Dossiers"] = fallback_stats["Dossiers"].fillna(0).astype(int)
fallback_stats["Réassignés"] = fallback_stats["Réassignés"].fillna(0).astype(int)
fallback_stats["Part du volume IA"] = (
    fallback_stats["Dossiers"] / n_cases * 100
).map(fmt_pct)
fallback_stats["Taux de réassignation"] = fallback_stats["Taux"].map(fmt_pct)
display(Markdown("### Fallback — Other et No detection"))
display(fallback_stats.rename_axis("Équipe").reset_index()[
    ["Équipe", "Dossiers", "Réassignés", "Part du volume IA", "Taux de réassignation"]
])

# BLOC 2 — AHT, heures, résolution directe et ping-pong
display(Markdown("## Bloc 2 — Efficacité opérationnelle et coût"))

case_reassigned = df_case["Is_Reassigned"]
case_well = ~case_reassigned
aht_well = df_case.loc[case_well, "Handle_Time_Min"].mean()
aht_reassigned = df_case.loc[case_reassigned, "Handle_Time_Min"].mean()
aht_penalty = aht_reassigned - aht_well
aht_penalty_pct = (
    aht_penalty / aht_well * 100
    if pd.notna(aht_well) and aht_well != 0 else np.nan
)
hours_net = n_case_reassigned * aht_penalty / 60

display(pd.DataFrame([
    ("AHT moyen — dossiers bien routés", f"{fmt_num(aht_well)} min"),
    ("AHT moyen — dossiers réassignés", f"{fmt_num(aht_reassigned)} min"),
    ("Pénalité AHT", f"{aht_penalty:+,.2f} min ({aht_penalty_pct:+.2f}%)"
     if pd.notna(aht_penalty_pct) else "N/D"),
    ("Heures-agent nettes", f"{hours_net:+,.2f} h",
     "Valeur négative = gain net d'AHT"),
], columns=["Indicateur", "Valeur", "Interprétation"]))

macro = df_case["Reassign_Macros"]
macro_valid = macro.notna() & macro.ge(0)
macro_count = int(macro_valid.sum())
direct_count = int((macro_valid & macro.eq(0)).sum())
direct_rate = direct_count / macro_count * 100 if macro_count else np.nan
display(Markdown("### Résolution directe — zéro macro de réassignation"))
display(pd.DataFrame([{
    "Dossiers à 0 macro": f"{direct_count:,}",
    "Dossiers avec macros exploitables": f"{macro_count:,}",
    "Taux de résolution directe": fmt_pct(direct_rate),
    "Macros manquantes / invalides": f"{int((~macro_valid).sum()):,}",
}]))

macro_bucket = pd.Series("Manquant / invalide", index=df_case.index, dtype="object")
macro_bucket.loc[macro_valid & macro.eq(0)] = "0"
for value in [1, 2, 3, 4]:
    macro_bucket.loc[macro_valid & macro.eq(value)] = str(value)
macro_bucket.loc[macro_valid & macro.ge(5)] = "5+"
bucket_order = ["0", "1", "2", "3", "4", "5+", "Manquant / invalide"]
ping_pong = (
    macro_bucket.value_counts()
    .reindex(bucket_order, fill_value=0)
    .rename_axis("Macros de transfert")
    .rename("Dossiers")
    .reset_index()
)
ping_pong["Part des dossiers"] = (ping_pong["Dossiers"] / n_cases * 100).map(fmt_pct)
display(Markdown("### Distribution exacte du ping-pong par dossier"))
display(ping_pong)

# BLOC 3 — CSAT, semaine, priorité, pays et Suisse
display(Markdown("## Bloc 3 — Impact client, temporalité et business"))

csat_well = df_case.loc[case_well, "CSAT_Initial"].mean()
csat_reassigned = df_case.loc[case_reassigned, "CSAT_Initial"].mean()
csat_drop = csat_well - csat_reassigned
display(pd.DataFrame([
    ("CSAT initial — bien routés", fmt_num(csat_well)),
    ("CSAT initial — réassignés", fmt_num(csat_reassigned)),
    ("Chute bien routés - réassignés", f"{csat_drop:+,.2f} points"),
], columns=["Indicateur", "Valeur"]))

# Une observation par dossier et par semaine, pour éviter de compter plusieurs
# lignes du même dossier plusieurs fois dans le volume hebdomadaire.
case_week = (
    df_line.loc[df_line["_Case_ID"].notna()]
    .groupby(["_Case_ID", "_Period_Label"], as_index=False)
    .agg(
        Period_Start=("_Period_Start", "min"),
        Is_Reassigned=("Is_Reassigned", "max"),
        CSAT_Initial=("CSAT_Initial", "first"),
        Handle_Time_Min=("Handle_Time_Min", "max"),
    )
)

weekly_rows = []
for period, group in case_week.groupby("_Period_Label", sort=False):
    well_week = group.loc[~group["Is_Reassigned"]]
    reassigned_week = group.loc[group["Is_Reassigned"]]
    csat_w = well_week["CSAT_Initial"].mean()
    csat_r = reassigned_week["CSAT_Initial"].mean()
    aht_w = well_week["Handle_Time_Min"].mean()
    aht_r = reassigned_week["Handle_Time_Min"].mean()
    weekly_rows.append({
        "Semaine": period,
        "Dossiers": len(group),
        "Réassignés": int(group["Is_Reassigned"].sum()),
        "Taux réassignation": fmt_pct(group["Is_Reassigned"].mean() * 100),
        "CSAT bien routés": fmt_num(csat_w),
        "CSAT réassignés": fmt_num(csat_r),
        "Écart CSAT (bien - réassignés)": (
            f"{csat_w - csat_r:+,.2f}" if pd.notna(csat_w) and pd.notna(csat_r) else "N/D"
        ),
        "AHT bien routés (min)": fmt_num(aht_w),
        "AHT réassignés (min)": fmt_num(aht_r),
        "Pénalité AHT (min)": (
            f"{aht_r - aht_w:+,.2f}" if pd.notna(aht_w) and pd.notna(aht_r) else "N/D"
        ),
        "_Sort": group["Period_Start"].min(),
    })
weekly_table = pd.DataFrame(weekly_rows).sort_values("_Sort").drop(columns="_Sort")
display(Markdown("### Analyse temporelle hebdomadaire — CSAT, AHT et pénalité"))
display(weekly_table)

priority_order = ["Low", "Medium", "High", "Immediate"]
priority_data = (
    df_case.groupby("_Priority")
    .agg(Dossiers=("Is_Reassigned", "size"),
         Réassignés=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
)
priority_order += [x for x in priority_data.index if x not in priority_order]
priority_data = priority_data.reindex(priority_order)
priority_data["Dossiers"] = priority_data["Dossiers"].fillna(0).astype(int)
priority_data["Réassignés"] = priority_data["Réassignés"].fillna(0).astype(int)
priority_data["Taux mauvais routage"] = priority_data["Taux"].map(fmt_pct)
display(Markdown("### Taux de mauvais routage par priorité"))
display(priority_data.rename_axis("Priorité").reset_index()[
    ["Priorité", "Dossiers", "Réassignés", "Taux mauvais routage"]
])

country_data = (
    df_case.groupby("_Country")
    .agg(Dossiers=("Is_Reassigned", "size"),
         Réassignés=("Is_Reassigned", "sum"),
         Taux=("Is_Reassigned", "mean"))
    .sort_values("Dossiers", ascending=False)
)
country_data["Taux mauvais routage"] = country_data["Taux"].map(fmt_pct)
display(Markdown("### Taux de mauvais routage par pays / marché"))
display(country_data.rename_axis("Pays / marché").reset_index()[
    ["Pays / marché", "Dossiers", "Réassignés", "Taux mauvais routage"]
])

swiss = df_case.loc[df_case["_Country"].str.casefold().isin(["switzerland", "suisse"])]
swiss_b2b = swiss.loc[swiss["_Initial_Team"].str.casefold().eq("b2b")]
swiss_b2b_correct = swiss_b2b["_Final_Category"].str.casefold().isin(valid_b2b_categories)
swiss_b2b_false = swiss_b2b.loc[~swiss_b2b_correct]
swiss_errors = int(swiss["Is_Reassigned"].sum())
swiss_false_b2b_errors = int(swiss_b2b_false["Is_Reassigned"].sum())
display(Markdown("### Focus Suisse & B2B"))
display(pd.DataFrame([
    ("Dossiers suisses", f"{len(swiss):,}"),
    ("Taux de réassignation suisse", fmt_pct(swiss["Is_Reassigned"].mean() * 100)),
    ("Dossiers qualifiés B2B en Suisse", f"{len(swiss_b2b):,}"),
    ("Précision B2B suisse",
     fmt_pct(swiss_b2b_correct.mean() * 100 if len(swiss_b2b) else np.nan)),
    ("Faux B2B suisses réassignés / erreurs suisses",
     f"{swiss_false_b2b_errors:,} ({fmt_pct(swiss_false_b2b_errors / swiss_errors * 100 if swiss_errors else np.nan)})"),
], columns=["Indicateur", "Valeur"]))

# BLOC 4 — flux croisés par dossier et matrice ligne d'export
display(Markdown("## Bloc 4 — Matrice de confusion et flux croisés"))

case_flows = df_case.loc[df_case["Is_Reassigned"]].groupby(
    ["_Initial_Team", "_Current_Team"]
).agg(Volume=("_Case_ID", "nunique")).reset_index()
line_flows = df_line.loc[df_line["Is_Reassigned"]].groupby(
    ["_Initial_Team", "_Current_Team"]
).size().rename("Lignes export").reset_index()
top_flows = (
    case_flows.merge(line_flows, on=["_Initial_Team", "_Current_Team"], how="left")
    .rename(columns={
        "_Initial_Team": "Équipe initiale",
        "_Current_Team": "Équipe finale observée",
    })
    .sort_values("Volume", ascending=False)
    .head(15)
)
display(Markdown("### Top 15 des flux d'erreurs — volume en dossiers distincts"))
display(top_flows.reset_index(drop=True))

# Matrice de confusion demandée en grain ligne d'export.
confusion_counts = pd.crosstab(
    df_line["_Initial_Team"],
    df_line["_Final_Category"],
).reindex(team_order, fill_value=0)
confusion_counts.index.name = "Équipe initiale IA"
confusion_counts.columns.name = "Catégorie finale réelle"
display(Markdown("### Matrice complète — volumes ligne d'export"))
display(confusion_counts)

confusion_pct = confusion_counts.div(
    confusion_counts.sum(axis=1).replace(0, np.nan), axis=0
) * 100
confusion_pct = confusion_pct.applymap(
    lambda value: fmt_pct(value) if pd.notna(value) else "N/D"
)
display(Markdown("### Matrice complète — % au sein de chaque équipe initiale"))
display(confusion_pct)

print(
    "\nRègles d'agrégation dossier : réassigné si au moins une ligne indique Yes; "
    "équipe initiale = première valeur observée; équipe courante et catégorie finale "
    "= dernières valeurs non vides après tri par période puis ordre export; "
    "AHT et macros = maximum observé par dossier. "
    "L'analyse hebdomadaire compte chaque dossier une fois par semaine."
)
```

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# 1. Chargement et validation des en-têtes demandés
file_path = Path("/Users/eloilahonde/Downloads/Analsye intent/CopyofClosedCasescount.xlsx")
if not file_path.is_file():
    raise FileNotFoundError(f"Fichier introuvable : {file_path}")

source_df = pd.read_excel(file_path, header=2)
source_df.columns = source_df.columns.astype(str).str.strip()

required = [
    "Routed by AI (Case)", "Initial Team (Case)",
    "[CORE] Teams Updated?? (Case)", "[CORE] Current Teams (Case)",
    "Level 1 - Case Category (Case)", "[CORE] Country (CF) (Case)",
    "Priority (Case)", "Handle Time (Case Clock) (SUM)",
    "Initial CSAT Score (Case)", "Reassignment Macro Usage Count (SUM)",
    "Case Details", "Date",
]
missing = [column for column in required if column not in source_df.columns]
if missing:
    raise KeyError(
        f"Colonnes absentes avec header=0 : {missing}. "
        f"Colonnes lues : {source_df.columns.tolist()}"
    )

# Garder seulement les lignes routées par l'IA.
ai_mask = (
    source_df["Routed by AI (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)
df_line = source_df.loc[ai_mask].copy()
if df_line.empty:
    raise ValueError("Aucune ligne avec Routed by AI (Case) = Yes.")

df_line["_Export_Row"] = np.arange(len(df_line))
df_line["_Case_ID"] = (
    df_line["Case Details"].astype("string").str.extract(r"^\s*(\d+)")[0]
)
df_line["Is_Reassigned"] = (
    df_line["[CORE] Teams Updated?? (Case)"]
    .astype("string").str.strip().str.casefold().eq("yes")
    .fillna(False)
)

def clean_label(value):
    if pd.isna(value) or str(value).strip().casefold() in {
        "", "-", "nan", "none", "<na>", "non renseigné"
    }:
        return "Non renseigné"
    return str(value).strip()

def get_core_domain(team_name):
    """Supprime le préfixe pays et harmonise les noms de domaine."""
    if pd.isna(team_name) or str(team_name).strip().casefold() in {
        "", "-", "nan", "none", "<na>", "non renseigné"
    }:
        return "Non renseigné"

    clean = re.sub(
        r"^[A-Za-z]{2}\s*-\s*",
        "",
        str(team_name).strip(),
    ).strip()

    domain_map = {
        "aftersale": "Aftersales",
        "aftersales": "Aftersales",
        "marketplace": "Marketplace",
        "presale": "Presales",
        "presales": "Presales",
        "b2b": "B2B",
        "services": "Services",
        "legal": "Legal",
        "other": "Other",
        "no detection": "No detection",
    }
    return domain_map.get(clean.casefold(), clean)

def label_domain_row(row):
    initial = row["Domain_Initial"]
    current = row["Domain_Current"]

    if initial in {"Other", "No detection"}:
        return "3. Fallback / Non détecté"
    if initial == "Non renseigné" or current == "Non renseigné":
        return "4. Domaine non évaluable"
    if initial == current:
        return "1. Routage IA Correct (Même Domaine)"
    return "2. Erreur de Routage IA (Changement de Domaine)"

def fmt_num(value, decimals=2):
    return "N/D" if pd.isna(value) else f"{value:,.{decimals}f}"

def fmt_pct(value):
    return "N/D" if pd.isna(value) else f"{value:.2f}%"

df_line["_Initial_Team_Raw"] = df_line["Initial Team (Case)"].map(clean_label)
df_line["_Current_Team_Raw"] = df_line["[CORE] Current Teams (Case)"].map(clean_label)
df_line["_Final_Category"] = df_line["Level 1 - Case Category (Case)"].map(clean_label)
df_line["_Country"] = df_line["[CORE] Country (CF) (Case)"].map(clean_label)
df_line["_Priority"] = df_line["Priority (Case)"].map(clean_label)

df_line["Domain_Initial"] = df_line["Initial Team (Case)"].map(get_core_domain)
df_line["Domain_Current"] = df_line["[CORE] Current Teams (Case)"].map(get_core_domain)
df_line["Routing_Quality"] = df_line.apply(label_domain_row, axis=1)
df_line["Is_Domain_Misroute"] = (
    df_line["Domain_Initial"].ne(df_line["Domain_Current"])
    & df_line["Domain_Initial"].ne("Non renseigné")
    & df_line["Domain_Current"].ne("Non renseigné")
    & ~df_line["Domain_Initial"].isin(["Other", "No detection"])
)

# Durée : les valeurs numériques Excel sont des jours; les Timedelta sont convertis directement.
raw_time = df_line["Handle Time (Case Clock) (SUM)"]
if pd.api.types.is_timedelta64_dtype(raw_time.dtype):
    df_line["Handle_Time_Min"] = raw_time.dt.total_seconds() / 60
else:
    numeric_days = pd.to_numeric(raw_time, errors="coerce")
    parsed_duration = pd.to_timedelta(
        raw_time.where(numeric_days.isna()), errors="coerce"
    )
    df_line["Handle_Time_Min"] = numeric_days * 24 * 60
    duration_rows = numeric_days.isna()
    df_line.loc[duration_rows, "Handle_Time_Min"] = (
        parsed_duration.loc[duration_rows].dt.total_seconds() / 60
    )

df_line["CSAT_Initial"] = pd.to_numeric(
    df_line["Initial CSAT Score (Case)"], errors="coerce"
)
df_line["Reassign_Macros"] = pd.to_numeric(
    df_line["Reassignment Macro Usage Count (SUM)"], errors="coerce"
)

# Date : l'export contient des périodes hebdomadaires; on en extrait le début.
period_start_text = (
    df_line["Date"].astype("string")
    .str.extract(r"([A-Za-z]+\s+\d{1,2}\s*,?\s+\d{4})")[0]
    .str.replace(r"\s+,", ",", regex=True)
)
df_line["_Period_Start"] = pd.to_datetime(period_start_text, errors="coerce")
df_line["_Period_Label"] = df_line["_Period_Start"].dt.strftime("%Y-%m-%d")
df_line["_Period_Label"] = df_line["_Period_Label"].fillna("Période non analysable")

# 2. Agrégation dossier : première équipe, dernière équipe/catégorie observées.
def first_value(series):
    values = [v for v in series if pd.notna(v) and v != "Non renseigné"]
    return values[0] if values else "Non renseigné"

def last_value(series):
    values = [v for v in series if pd.notna(v) and v != "Non renseigné"]
    return values[-1] if values else "Non renseigné"

ordered_lines = df_line.sort_values(
    ["_Case_ID", "_Period_Start", "_Export_Row"],
    kind="stable",
    na_position="last",
)

case_rows = []
identified_lines = ordered_lines.loc[ordered_lines["_Case_ID"].notna()]
for case_id, group in identified_lines.groupby("_Case_ID", sort=False):
    group = group.sort_values(
        ["_Period_Start", "_Export_Row"],
        kind="stable",
        na_position="last",
    )
    aht_values = group["Handle_Time_Min"].dropna()
    macro_values = group["Reassign_Macros"].dropna()
    csat_values = group["CSAT_Initial"].dropna()

    record = {
        "_Case_ID": case_id,
        "Initial_Team": first_value(group["_Initial_Team_Raw"]),
        "Current_Team": last_value(group["_Current_Team_Raw"]),
        "Final_Category": last_value(group["_Final_Category"]),
        "Country": first_value(group["_Country"]),
        "Priority": first_value(group["_Priority"]),
        "Is_Reassigned": bool(group["Is_Reassigned"].any()),
        "Handle_Time_Min": aht_values.max() if len(aht_values) else np.nan,
        "Reassign_Macros": macro_values.max() if len(macro_values) else np.nan,
        "CSAT_Initial": csat_values.iloc[0] if len(csat_values) else np.nan,
        "Export_Rows": len(group),
    }
    record["Domain_Initial"] = get_core_domain(record["Initial_Team"])
    record["Domain_Current"] = get_core_domain(record["Current_Team"])
    record["Routing_Quality"] = label_domain_row(record)
    record["Is_Domain_Misroute"] = (
        record["Domain_Initial"] != record["Domain_Current"]
        and record["Domain_Initial"] not in {"Other", "No detection", "Non renseigné"}
        and record["Domain_Current"] != "Non renseigné"
    )
    case_rows.append(record)

df_case = pd.DataFrame(case_rows)
if df_case.empty:
    raise ValueError("Aucun ID dossier numérique n'a été extrait de Case Details.")

all_cases = len(df_case)
all_lines = len(df_line)
fallback_names = {"Other", "No detection"}
evaluable_cases = ~df_case["Domain_Initial"].isin(fallback_names | {"Non renseigné"})
evaluable_cases &= df_case["Domain_Current"].ne("Non renseigné")

# BLOC 1 — qualité du routage normalisée par domaine
display(Markdown("## Bloc 1 — Précision et qualité du routage IA corrigées"))

display(pd.DataFrame([
    ("Réassignation brute Sprinklr — dossiers",
     fmt_pct(df_case["Is_Reassigned"].mean() * 100)),
    ("Vraie erreur de domaine — dossiers / tous les dossiers",
     fmt_pct(df_case["Is_Domain_Misroute"].mean() * 100)),
    ("Vraie erreur de domaine — dossiers évaluables",
     fmt_pct(df_case.loc[evaluable_cases, "Is_Domain_Misroute"].mean() * 100)),
    ("Réassignation brute Sprinklr — lignes",
     fmt_pct(df_line["Is_Reassigned"].mean() * 100)),
    ("Vraie erreur de domaine — lignes / toutes les lignes",
     fmt_pct(df_line["Is_Domain_Misroute"].mean() * 100)),
    ("Dossiers évaluables / dossiers totaux",
     f"{int(evaluable_cases.sum()):,} / {all_cases:,}"),
], columns=["Indicateur", "Valeur"]))

domain_order = [
    "Aftersales", "Marketplace", "Presales", "B2B",
    "Legal", "Services", "Other", "No detection",
]
domain_order += [
    value for value in df_case["Domain_Initial"].unique()
    if value not in domain_order
]

def domain_error_table(data, team_col, grain):
    grouped = data.groupby(team_col).agg(
        Volume=("Is_Domain_Misroute", "size"),
        Erreurs=("Is_Domain_Misroute", "sum"),
        Taux=("Is_Domain_Misroute", "mean"),
    )
    grouped = grouped.reindex(domain_order)
    grouped["Volume"] = grouped["Volume"].fillna(0).astype(int)
    grouped["Erreurs"] = grouped["Erreurs"].fillna(0).astype(int)
    grouped["Taux vraie erreur domaine"] = grouped["Taux"].map(fmt_pct)
    grouped["Grain"] = grain
    return grouped.rename_axis("Domaine initial").reset_index()[
        ["Grain", "Domaine initial", "Volume", "Erreurs", "Taux vraie erreur domaine"]
    ]

display(Markdown("### Taux d'erreur de domaine par équipe initiale"))
display(pd.concat([
    domain_error_table(df_case, "Domain_Initial", "Dossier"),
    domain_error_table(df_line, "Domain_Initial", "Ligne export"),
], ignore_index=True))

b2b_cases = df_case.loc[df_case["Domain_Initial"].eq("B2B")]
b2b_correct = b2b_cases["Domain_Current"].eq("B2B")
b2b_precision = b2b_correct.mean() * 100 if len(b2b_cases) else np.nan
b2b_false = b2b_cases.loc[~b2b_correct & b2b_cases["Domain_Current"].ne("Non renseigné")]

display(Markdown("### Précision B2B révisée — même domaine fonctionnel"))
display(pd.DataFrame([
    ("Dossiers initialement B2B", f"{len(b2b_cases):,}"),
    ("Restés dans le domaine B2B", f"{int(b2b_correct.sum()):,} ({fmt_pct(b2b_precision)})"),
    ("Faux positifs réorientés vers un autre domaine", f"{len(b2b_false):,}"),
], columns=["Indicateur", "Valeur"]))

if len(b2b_false):
    display(
        b2b_false.groupby("Domain_Current")
        .size().rename("Dossiers").reset_index()
        .rename(columns={"Domain_Current": "Domaine final des faux B2B"})
        .sort_values("Dossiers", ascending=False)
    )

fallback = df_case.loc[df_case["Domain_Initial"].isin(fallback_names)]
fallback_table = (
    fallback.groupby("Domain_Initial")
    .agg(Dossiers=("Is_Reassigned", "size"),
         Réassignés=("Is_Reassigned", "sum"),
         Taux_réassignation=("Is_Reassigned", "mean"))
    .reindex(["Other", "No detection"])
)
fallback_table["Dossiers"] = fallback_table["Dossiers"].fillna(0).astype(int)
fallback_table["Réassignés"] = fallback_table["Réassignés"].fillna(0).astype(int)
fallback_table["Taux réassignation"] = fallback_table["Taux_réassignation"].map(fmt_pct)
display(Markdown("### Fallback — volume et réassignation brute"))
display(fallback_table.rename_axis("Domaine initial").reset_index()[
    ["Domaine initial", "Dossiers", "Réassignés", "Taux réassignation"]
])

# BLOC 2 — coûts et macros au grain dossier
display(Markdown("## Bloc 2 — Efficacité opérationnelle et coût"))

correct_cases = df_case["Routing_Quality"].eq("1. Routage IA Correct (Même Domaine)")
error_cases = df_case["Is_Domain_Misroute"]
aht_correct = df_case.loc[correct_cases, "Handle_Time_Min"].mean()
aht_error = df_case.loc[error_cases, "Handle_Time_Min"].mean()
aht_penalty = aht_error - aht_correct
aht_penalty_pct = (
    aht_penalty / aht_correct * 100
    if pd.notna(aht_correct) and aht_correct != 0 else np.nan
)
hours_net = int(error_cases.sum()) * aht_penalty / 60

display(pd.DataFrame([
    ("AHT moyen — même domaine", f"{fmt_num(aht_correct)} min"),
    ("AHT moyen — vraie erreur de domaine", f"{fmt_num(aht_error)} min"),
    ("Pénalité AHT", f"{aht_penalty:+,.2f} min ({aht_penalty_pct:+.2f}%)"
     if pd.notna(aht_penalty_pct) else "N/D"),
    ("Heures-agent nettes estimées", f"{hours_net:+,.2f} h"),
], columns=["Indicateur", "Valeur"]))

macro = df_case["Reassign_Macros"]
macro_valid = macro.notna() & macro.ge(0)
direct_count = int((macro_valid & macro.eq(0)).sum())
direct_rate = direct_count / int(macro_valid.sum()) * 100 if macro_valid.any() else np.nan
display(pd.DataFrame([{
    "Dossiers avec 0 macro": f"{direct_count:,}",
    "Dossiers avec macro exploitable": f"{int(macro_valid.sum()):,}",
    "Taux résolution directe": fmt_pct(direct_rate),
    "Macros manquantes / invalides": f"{int((~macro_valid).sum()):,}",
}]))

macro_bucket = pd.Series("Manquant / invalide", index=df_case.index, dtype="object")
macro_bucket.loc[macro_valid & macro.eq(0)] = "0"
for value in [1, 2, 3, 4]:
    macro_bucket.loc[macro_valid & macro.eq(value)] = str(value)
macro_bucket.loc[macro_valid & macro.ge(5)] = "5+"
ping_pong = (
    macro_bucket.value_counts()
    .reindex(["0", "1", "2", "3", "4", "5+", "Manquant / invalide"], fill_value=0)
    .rename_axis("Macros de transfert")
    .rename("Dossiers")
    .reset_index()
)
ping_pong["Part des dossiers"] = (ping_pong["Dossiers"] / all_cases * 100).map(fmt_pct)
display(Markdown("### Distribution du ping-pong par dossier"))
display(ping_pong)

# BLOC 3 — CSAT, temporalité, priorité, pays et Suisse
display(Markdown("## Bloc 3 — Impact client et analyse temporelle"))

csat_correct = df_case.loc[correct_cases, "CSAT_Initial"].mean()
csat_error = df_case.loc[error_cases, "CSAT_Initial"].mean()
display(pd.DataFrame([
    ("CSAT initial — même domaine", fmt_num(csat_correct)),
    ("CSAT initial — erreur de domaine", fmt_num(csat_error)),
    ("Chute de CSAT (correct - erreur)", f"{csat_correct - csat_error:+,.2f} points"),
], columns=["Indicateur", "Valeur"]))

# Une observation par dossier/semaine; AHT = maximum observé dans cette semaine.
case_week = (
    df_line.loc[df_line["_Case_ID"].notna()]
    .groupby(["_Case_ID", "_Period_Label"], as_index=False)
    .agg(
        Period_Start=("_Period_Start", "min"),
        Domain_Initial=("Domain_Initial", "first"),
        Domain_Current=("Domain_Current", "last"),
        Is_Reassigned=("Is_Reassigned", "max"),
        Handle_Time_Min=("Handle_Time_Min", "max"),
        CSAT_Initial=("CSAT_Initial", "first"),
    )
)
case_week["Is_Domain_Misroute"] = (
    case_week["Domain_Initial"].ne(case_week["Domain_Current"])
    & case_week["Domain_Initial"].ne("Non renseigné")
    & case_week["Domain_Current"].ne("Non renseigné")
    & ~case_week["Domain_Initial"].isin(fallback_names)
)

weekly_rows = []
for period, group in case_week.groupby("_Period_Label", sort=False):
    correct_week = group.loc[~group["Is_Domain_Misroute"]]
    error_week = group.loc[group["Is_Domain_Misroute"]]
    csat_c = correct_week["CSAT_Initial"].mean()
    csat_e = error_week["CSAT_Initial"].mean()
    aht_c = correct_week["Handle_Time_Min"].mean()
    aht_e = error_week["Handle_Time_Min"].mean()
    weekly_rows.append({
        "Semaine": period,
        "Dossiers": len(group),
        "Erreurs domaine": int(group["Is_Domain_Misroute"].sum()),
        "Taux erreur domaine": fmt_pct(group["Is_Domain_Misroute"].mean() * 100),
        "CSAT même domaine": fmt_num(csat_c),
        "CSAT erreur domaine": fmt_num(csat_e),
        "Écart CSAT": f"{csat_c - csat_e:+,.2f}" if pd.notna(csat_c) and pd.notna(csat_e) else "N/D",
        "AHT même domaine (min)": fmt_num(aht_c),
        "AHT erreur domaine (min)": fmt_num(aht_e),
        "Pénalité AHT (min)": f"{aht_e - aht_c:+,.2f}" if pd.notna(aht_c) and pd.notna(aht_e) else "N/D",
        "_Sort": group["Period_Start"].min(),
    })
weekly_table = pd.DataFrame(weekly_rows).sort_values("_Sort").drop(columns="_Sort")
display(Markdown("### Analyse hebdomadaire — CSAT, AHT et pénalité"))
display(weekly_table)

def grouped_domain_error(data, column):
    out = data.groupby(column).agg(
        Volume=("Is_Domain_Misroute", "size"),
        Erreurs=("Is_Domain_Misroute", "sum"),
        Taux=("Is_Domain_Misroute", "mean"),
    )
    out["Taux vraie erreur domaine"] = out["Taux"].map(fmt_pct)
    return out.rename_axis(column).reset_index()[
        [column, "Volume", "Erreurs", "Taux vraie erreur domaine"]
    ]

display(Markdown("### Erreur de domaine par priorité"))
display(grouped_domain_error(df_case, "Priority"))

display(Markdown("### Erreur de domaine par pays"))
display(grouped_domain_error(df_case, "Country").sort_values("Volume", ascending=False))

swiss = df_case.loc[df_case["Country"].str.casefold().isin(["switzerland", "suisse"])]
swiss_b2b = swiss.loc[swiss["Domain_Initial"].eq("B2B")]
swiss_b2b_false = swiss_b2b.loc[swiss_b2b["Is_Domain_Misroute"]]
swiss_errors = int(swiss["Is_Domain_Misroute"].sum())
swiss_b2b_error_share = (
    len(swiss_b2b_false) / swiss_errors * 100 if swiss_errors else np.nan
)
display(Markdown("### Focus Suisse & B2B révisé"))
display(pd.DataFrame([
    ("Dossiers suisses", f"{len(swiss):,}"),
    ("Taux vraie erreur de domaine suisse",
     fmt_pct(swiss["Is_Domain_Misroute"].mean() * 100)),
    ("Dossiers qualifiés B2B en Suisse", f"{len(swiss_b2b):,}"),
    ("Précision B2B suisse",
     fmt_pct((swiss_b2b["Domain_Current"].eq("B2B").mean() * 100)
             if len(swiss_b2b) else np.nan)),
    ("Faux B2B suisses / erreurs de domaine suisses",
     f"{len(swiss_b2b_false):,} ({fmt_pct(swiss_b2b_error_share)})"),
], columns=["Indicateur", "Valeur"]))

# BLOC 4 — flux par domaines et taxonomie réelle
display(Markdown("## Bloc 4 — Matrice de confusion et flux croisés"))

domain_flows = (
    df_case.loc[df_case["Is_Domain_Misroute"]]
    .groupby(["Domain_Initial", "Domain_Current"])
    .agg(Volume=("_Case_ID", "nunique"))
    .reset_index()
    .rename(columns={
        "Domain_Initial": "Domaine initial",
        "Domain_Current": "Domaine final",
    })
    .sort_values("Volume", ascending=False)
    .head(15)
)
display(Markdown("### Top 15 des vrais flux d'erreurs entre domaines"))
display(domain_flows)

# Matrice gardée au grain ligne d'export, comme demandé.
confusion_counts = pd.crosstab(
    df_line["Domain_Initial"],
    df_line["_Final_Category"],
)
confusion_counts.index.name = "Domaine initial"
confusion_counts.columns.name = "Catégorie finale réelle"
display(Markdown("### Matrice domaine initial × catégorie finale — volumes ligne export"))
display(confusion_counts)

confusion_pct = confusion_counts.div(
    confusion_counts.sum(axis=1).replace(0, np.nan), axis=0
) * 100
confusion_pct = confusion_pct.applymap(
    lambda value: fmt_pct(value) if pd.notna(value) else "N/D"
)
display(Markdown("### Matrice domaine initial × catégorie finale — % par domaine initial"))
display(confusion_pct)

print(
    "Règles : fallback Other/No detection exclu du taux de vraie erreur; "
    "domaine final manquant classé non évaluable; AHT et macros au dossier = "
    "maximum observé; matrice et flux de parcours conservés à partir des lignes."
)
```
